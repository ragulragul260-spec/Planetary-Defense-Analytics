```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")

print("✅ All Libraries Imported Successfully")
```

```python
df = pd.read_csv("../Data/raw/pha_dataset.csv")

df.head()
```

```python
print("Rows, Columns :", df.shape)

df.head()
```

```python
df.info()
```

```python
df.describe().T
```

```python
missing = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_%": (df.isnull().sum()/len(df))*100
})

missing = missing[missing["Missing_Count"] > 0]

missing.sort_values("Missing_%", ascending=False)
```

```python
print("Duplicate Rows :", df.duplicated().sum())
```

```python
identifier_cols = [
    'id',
    'spkid',
    'full_name',
    'pdes',
    'name',
    'orbit_id'
]

df.drop(columns=identifier_cols,
        inplace=True,
        errors='ignore')
```

```python
sigma_cols = [col for col in df.columns
              if "sigma" in col.lower()]

sigma_cols
```

```python
df.drop(columns=sigma_cols,
        inplace=True)
```

```python
print("Sigma Columns Removed :", len(sigma_cols))
```

```python
zero_var_cols = []

for col in df.columns:
    if df[col].nunique() == 1:
        zero_var_cols.append(col)

zero_var_cols
```

```python
df.drop(columns=zero_var_cols,
        inplace=True)
```

```python
unique_cols = []

for col in df.columns:

    if df[col].nunique() == len(df):
        unique_cols.append(col)

unique_cols
```

```python
df.drop(columns=unique_cols,
        errors="ignore",
        inplace=True)
```

```python
print(df.shape)
```

```python
df.columns.tolist()
```

```python
print(
    "Negative e values :",
    (df['e'] < 0).sum()
)
```

```python
print(
    "Invalid a values :",
    (df['a'] <= 0).sum()
)
```

```python
print(
    "Negative diameter :",
    (df['diameter'] <= 0).sum()
)
```

```python
print(
    "Negative albedo :",
    (df['albedo'] < 0).sum()
)
```

```python
df = df[df['e'] >= 0]

df = df[df['a'] > 0]

df = df[
    (df['diameter'].isna())
    |
    (df['diameter'] > 0)
]

df = df[
    (df['albedo'].isna())
    |
    (df['albedo'] >= 0)
]
```

```python
print(df.shape)
```

```python
angle_cols = [
    'om',
    'w',
    'ma'
]

for col in angle_cols:

    radians = np.deg2rad(
        df[col]
    )

    df[col+'_sin'] = np.sin(
        radians
    )

    df[col+'_cos'] = np.cos(
        radians
    )
```

```python
df.head()
```

```python
df.drop(
    columns=[
        'om',
        'w',
        'ma'
    ],
    inplace=True
)
```

```python
classification_df = df[
    df['pha'].notna()
].copy()
```

```python
classification_df.shape
```

```python
classification_df.drop(
    columns=[
        'neo',
        'moid',
        'moid_ld',
        'class'
    ],
    inplace=True,
    errors='ignore'
)
```

```python
classification_df.shape
```

```python
regression_df = df[
    df['diameter'].notna()
].copy()
```

```python
regression_df.shape
```
